```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("C:/Users/DELL/Desktop/Social_Media_Engagement_Analysis/Dataset/Social Media Engagement Dataset.csv")

# Display first 5 rows
print(df.head())
```

```python
print(df.shape)
print(df.columns)
print(df.info())
```

```python
# Dataset size
print(df.shape)
```

```python
# Column names
print(df.columns)
```

```python
# Dataset information
print(df.info())
```

```python
# Statistical summary
print(df.describe())
```

```python
# Check missing values in each column
print(df.isnull().sum())
```

```python
# Handle Missing Values
df['mentions'] = df['mentions'].fillna('')
```

```python
# Check Duplicate Records
print("Duplicate Rows:", df.duplicated().sum())
```

```python
# Remove Duplicate Records
df = df.drop_duplicates()
```

```python
print(df.isnull().sum())
print("Duplicate Rows:", df.duplicated().sum())
```

```python
# Convert Timestamp into DateTime
df['timestamp'] = pd.to_datetime(df['timestamp'], errors='coerce')
```

```python
# Create Total Engagement Column
df['total_engagement'] = (
    df['likes_count'] +
    df['comments_count'] +
    df['shares_count']
)
print(df[['likes_count', 'comments_count','shares_count', 'total_engagement']].head())
```

```python
# Create Month Column
df['month'] = df['timestamp'].dt.month
print(df[['timestamp', 'month']].head())
```

```python
# Create Year Column
df['year'] = df['timestamp'].dt.year
print(df[['timestamp', 'year']].head())
```

```python
# Create Month Name Column
df['month_name'] = df['timestamp'].dt.month_name()
print(df[['timestamp', 'month_name']].head())
```

```python
# Check Updated Dataset
print(df.shape)
print(df.head())
```

```python
# Total Posts on Each Platform
platform_posts = df['platform'].value_counts()
print(platform_posts)
```

```python
# Platform-wise Total Engagement
platform_engagement = df.groupby('platform')['total_engagement'].sum()
print(platform_engagement)
```

```python
# Average Likes on Each Platform
avg_likes = df.groupby('platform')['likes_count'].mean()
print(avg_likes)
```

```python
# Average Comments on Each Platform
avg_comments = df.groupby('platform')['comments_count'].mean()
print(avg_comments)
```

```python
# Average Shares on Each Platform
avg_shares = df.groupby('platform')['shares_count'].mean()
print(avg_shares)
```

```python
# Top 10 Posts Based on Engagement
top_posts = df.sort_values(
    by='total_engagement',
    ascending=False
)
print(top_posts[['post_id', 'platform',
                 'total_engagement']].head(10))
```

```python
# Highest Impressions
top_impressions = df.sort_values(
    by='impressions',
    ascending=False
)
print(top_impressions[['post_id', 'platform',
                       'impressions']].head(10))
```

```python
# DATA VISUALIZATION

#  Platform-wise Total Posts
platform_posts = df['platform'].value_counts()
plt.figure(figsize=(8, 5))
platform_posts.plot(kind='bar')
plt.title('Total Posts by Platform')
plt.xlabel('Social Media Platform')
plt.ylabel('Number of Posts')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Platform-wise Total Engagement– Bar Chart

platform_engagement = df.groupby('platform')['total_engagement'].sum()
plt.figure(figsize=(8, 5))
platform_engagement.plot(kind='bar')
plt.title('Total Engagement by Platform')
plt.xlabel('Social Media Platform')
plt.ylabel('Total Engagement')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Platform-wise Posts Distribution

plt.figure(figsize=(7, 7))
platform_posts.plot(
    kind='pie',
    autopct='%1.1f%%',
    startangle=90
)
plt.title('Platform-wise Posts Distribution')
plt.ylabel('')
plt.show()
```

```python
# MONTHLY ENGAGEMENT ANALYSIS

monthly_posts = df.groupby('month_name')['post_id'].count()
print(monthly_posts)
```

```python
monthly_posts = df.groupby(
    df['timestamp'].dt.to_period('M')
)['post_id'].count()
print(monthly_posts)
```

```python
# Monthly Posts – Line Chart

plt.figure(figsize=(12, 5))
monthly_posts.plot(kind='line', marker='o')
plt.title('Monthly Total Posts Trend')
plt.xlabel('Month')
plt.ylabel('Total Posts')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Monthly Total Engagement

monthly_engagement = df.groupby(
    df['timestamp'].dt.to_period('M')
)['total_engagement'].sum()
print(monthly_engagement)
```

```python
# Monthly Engagement – Line Chart

plt.figure(figsize=(12, 5))
monthly_engagement.plot(kind='line', marker='o')
plt.title('Monthly Total Engagement Trend')
plt.xlabel('Month')
plt.ylabel('Total Engagement')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Check Sentiment Categories

sentiment_count = df['sentiment_label'].value_counts()
print(sentiment_count)
```

```python
# Sentiment Percentage

sentiment_percentage = df['sentiment_label'].value_counts(normalize=True) * 100
print(sentiment_percentage.round(2))
```

```python
# Sentiment Analysis – Bar Chart

plt.figure(figsize=(8, 5))
sentiment_count.plot(kind='bar')
plt.title('Sentiment Analysis')
plt.xlabel('Sentiment')
plt.ylabel('Number of Posts')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Sentiment Analysis – Pie Chart

plt.figure(figsize=(7, 7))
sentiment_count.plot(
    kind='pie',
    autopct='%1.1f%%',
    startangle=90
)
plt.title('Sentiment Distribution')
plt.ylabel('')
plt.show()
```

```python
# Sentiment-wise Average Engagement

sentiment_engagement = df.groupby(
    'sentiment_label'
)['total_engagement'].mean()
print(sentiment_engagement)
```

```python
# Check Hashtags

print(df['hashtags'].head(10))
```

```python
# Split Hashtags

hashtags = df['hashtags'].dropna().str.split(',')
print(hashtags.head())
```

```python
# Count Hashtag Frequency

from collections import Counter
all_hashtags = []
for row in hashtags:
    for tag in row:
        all_hashtags.append(tag.strip().lower())

hashtag_count = Counter(all_hashtags)
print(hashtag_count.most_common(10))
```

```python
# Create Top 10 Hashtags DataFrame

top_hashtags = pd.DataFrame(
    hashtag_count.most_common(10),
    columns=['Hashtag', 'Count']
)
print(top_hashtags)
```

```python
# Top 10 Hashtags – Bar Chart

plt.figure(figsize=(10, 5))
plt.bar(top_hashtags['Hashtag'], top_hashtags['Count'])
plt.title('Top 10 Trending Hashtags')
plt.xlabel('Hashtags')
plt.ylabel('Usage Count')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Check Campaign Names

print(df['campaign_name'].value_counts())
```

```python
# Campaign-wise Total Posts

campaign_posts = df.groupby('campaign_name')['post_id'].count()
print(campaign_posts)
```

```python
# Campaign-wise Total Engagement

campaign_engagement = df.groupby('campaign_name')['total_engagement'].sum()
print(campaign_engagement.sort_values(ascending=False))
```

```python
# Campaign-wise Average Engagement

campaign_avg = df.groupby('campaign_name')['total_engagement'].mean()
print(campaign_avg.sort_values(ascending=False))
```

```python
# Campaign Performance – Bar Chart

plt.figure(figsize=(12, 5))
campaign_engagement.sort_values(ascending=False).plot(kind='bar')
plt.title('Campaign-wise Total Engagement')
plt.xlabel('Campaign Name')
plt.ylabel('Total Engagement')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Campaign Phase Analysis

phase_engagement = df.groupby('campaign_phase')['total_engagement'].sum()
print(phase_engagement)
```

```python
# Platform-wise Sentiment Count

platform_sentiment = pd.crosstab(
    df['platform'],
    df['sentiment_label']
)
print(platform_sentiment)
```

```python
# Platform-wise Sentiment – Stacked Bar Chart

platform_sentiment.plot(
    kind='bar',
    stacked=True,
    figsize=(10, 5)
)
plt.title('Platform-wise Sentiment Analysis')
plt.xlabel('Social Media Platform')
plt.ylabel('Number of Posts')
plt.xticks(rotation=45)
plt.legend(title='Sentiment')
plt.tight_layout()
plt.show()
```

```python
# Platform-wise Average Engagement by Sentiment

platform_avg_engagement = df.pivot_table(
    index='platform',
    columns='sentiment_label',
    values='total_engagement',
    aggfunc='mean'
)
print(platform_avg_engagement.round(2))
```

```python
# Visualize Average Engagement

platform_avg_engagement.plot(
    kind='bar',
    figsize=(10, 5)
)
plt.title('Average Engagement by Platform and Sentiment')
plt.xlabel('Platform')
plt.ylabel('Average Engagement')
plt.xticks(rotation=45)
plt.legend(title='Sentiment')
plt.tight_layout()
plt.show()
```
